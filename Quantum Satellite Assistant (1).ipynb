#### Create a class that does the conversion of images to actual tensors.

```python
# now create a class (Dataset) that does the conversion of our images to actual tensors
class ImagePathDataset(Dataset):
    # so we want our class to take in the list of image paths, the list of labels and the transform function we have defined
    def __init__(self, paths, labels, transform):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        # here we would like to know the number of images
        return len(self.paths)

    def __getitem__(self, i):
        # now we open each image and convert it to RGB and apply our transforms, then we return the image and the label
        img = Image.open(self.paths[i]).convert("RGB")
        img = self.transform(img)
        return img, self.labels[i]
```

## Angle Encoding of the data

### Angle normalisation of the data

#### Create a function to angle normalise the data

```python
# we want to normalise our data over the angular range [-pi - pi], so since python has no defined function or class that can do this easily for us, we have to define the function for ourselves
class AngleNormalizer:
    # define the init method first
    def __init__(self):
        # so here we define the self parameter and the maximum variable per feature column
        self.max_abs_ = None
    def fit(self, X):
        # here we determine the absolute maximum value for each feature column
        self.max_abs_ = np.max(np.abs(X), axis = 0)
        # now just to make sure that we don't make mistake and get a zero maximum stored which can affect the model in the future
        # we assisgn every zero here to one
        self.max_abs_[self.max_abs_ == 0] = 1.0
        return self

    def transform(self, X):
        return np.pi * (X / self.max_abs_)

    def fit_transform(self, X):
        # in this method, we call th fit function on the data, store the maximum absolute values and return the normalised data from the trnsformed method
        self.fit(X)
        return self.transform(X)
```

## Let's build the Qnn circuit

### Define the QNN circuit

```python
# define the decorator and use autograd interface differentiator
@qml.qnode(dev, interface = "autograd")

# now let's define the circuit
def qnn(features, weights):
    # encode data
    for i in range(n_qubits):
        qml.RY(features[i], wires = i)

    # now the two trainable layers will be built using for loops
    for k in range(weights.shape[0]):
        for i in range(n_qubits):
            qml.Rot(
                weights[1, i, 0],
                weights[1, i, 1],
                weights[1, i, 2],
                wires = i
            )
        # now we entangle the qubits that have learned from the pca features
        for i in range(n_qubits - 1):
            qml.CNOT(wires = [i, i + 1])

    # now the QNN circuit will return a result which is the pauli expected value for 10 classes
    return [
        qml.expval(qml.PauliZ(0)),
        qml.expval(qml.PauliZ(1)),
        qml.expval(qml.PauliZ(2)),
        qml.expval(qml.PauliZ(3))
    ]
```

### Label encoding 

```python
# define a hot label encoding function
def one_hot(y, num_classes = 10):
    # create an array of zeros that will have this shape - the length of y ros and num of classes columns
    oh = np.zeros((len(y), num_classes))
    # now create an array of the lenght of y, and the next column should be the values of y then map it to 1
    oh[np.arange(len(y)), y] = 1
    return oh
```

### Define some crucial parameters and weights

```python
# define the number of classes
n_classes = len(listOfClasses)

# now define the number of outputs from our qnn circuit function
n_quantum_outputs = 4

# now we define some classical weights and bias we will make use of in our loss function
w_classical = np.random.randn(n_quantum_outputs, n_classes)
b_classical = np.zeros(n_classes)
```

### Define the loss function

```python
# here we define the loss function by calling our already defined qnn circuit and calating our prediction and the mean apsolute error of our prediction
# There is a numpy fnction we shall be using here which is pennylane's numpy for this work
import pennylane.numpy as pnp
def loss(weights, x, y):
    # create an array that stores the predictions of the qnn circuit from the train data
    predict = pnp.array([qnn(i, weights) for i in x])
    # now our results are in the form [-1, 1] we scale to the form [0, 1]
    predict = (predict + 1) / 2
    # so we defined the earlier weights because we want to be able to convert our 4 qnn outputs to classify our images among 10 classes
    # so we multiply our prediction by our classical weight then we add our bias
    logits = predict @ w_classical + b_classical
    # use softmax to create probabilities out of logits
    # to do that we first take the exponential of the logits subtracted from the largest logits
    exp_logits = pnp.exp(logits - pnp.max(logits, axis = 1, keepdims = True))
    # then we divide the exponential of the difference of each logits from the largest by the sum of them. THat is the probability
    probs = exp_logits / pnp.sum(exp_logits, axis = 1, keepdims = True)
    # using cross entropy to calculate our loss as we want to penalise the model for giving the right class a low probability
    # so we multiply out hot encoded labels to the log of the probability our model gave us
    return -pnp.mean(pnp.sum(y * pnp.log(probs + 1e-8), axis = 1))
```

```python
# so now the evaluation and testing function defined
def evaluate_qnn(x, y_true, weights):
    # get the predictions from the qnn circuit
    preds = pnp.array([qnn(i, weights) for i in x])
    # convert the predictions from the [-1, 1] form to [0, 1] form
    preds = (preds + 1) / 2
    # generate the logits from the predictions
    logits = preds @ w_classical + b_classical
    # use soft max to geerate probabilities from the logits of the predictions
    # first we generate the exponential of the logits substractd from the maximum logit per row
    exp_logits = pnp.exp(logits - pnp.max(logits, axis = 1, keepdims = True))
    # generate the probabilities per row now
    probs = exp_logits / pnp.sum(exp_logits, axis = 1, keepdims = True)
    # so we get the highest probability and asign it a class then store the class in a variable y_pred
    y_pred = pnp.argmax(probs, axis = 1)
    # calculate the accuracy of the model
    accuracy = pnp.mean(y_pred == y_true)
    # return the accuracy
    return accuracy
```

```python
###### so we want the training to be done in batches of 16 and for the training to  be done a 100 times
# so we import tqdm to help us keep track of time
from tqdm import tqdm
batch_size = 16
n_epochs = 100

#initialise a variable to be used to store the sum of the batch losses per epoch and the number of batches
sum_batch_loss = 0.0
num_batches = 0.0

# create a for loop to loop through each epoch
for epoch in range(n_epochs):
    # shuffle indices in the training data
    indices = np.random.permutation(len(x_train_encoded))
    # so now we want to loop through the training data in batches to do the training 
    for start in tqdm (range(0, len(x_train_encoded), batch_size), desc = f"Epoch {epoch + 1} / {n_epochs}"):
        batch_index = indices[start : start + batch_size]

        # now we use those shuffled indices we have gotten to select the x_train and y_train data to be used per batch
        x_batch = x_train_encoded[batch_index]
        y_batch = y_train[batch_index]

        # then the loss function takes in the weights, the batch of the x and y train and now learning begins
        weights = opt.step(lambda w_t: loss(w_t, x_batch, y_batch), weights)

        # now we calculate the loss per batch
        batch_loss = loss(weights, x_batch, y_batch)
        # then we add all the losses of each batch to get the loss of the whole epoch
        sum_batch_loss += batch_loss
        num_batches += 1

    # so here i will calculate the average batch loss
    avg_batch_loss = sum_batch_loss / num_batches
    # get the validation score of each epoch
    epoch_accuracy = evaluate_qnn(x_val_encoded, y_val, weights)

    # now print done when the epoch is done print done
    print(f"The training of Epoch {epoch + 1} is done ")
    print(f"The trainng loss at the end of {epoch + 1} is {avg_batch_loss:.4f}")
    print(f"The validation accuracy of {epoch + 1} is {epoch_accuracy:.4f}")
```
